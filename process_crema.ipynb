### use crema to label chords

use process_crema.yaml to get correct env

```python
import crema
from crema.models.chord import ChordModel

from pathlib import Path
import pretty_midi
import scipy.io.wavfile as wavfile
from IPython.display import Audio

def render_midi_to_wav(midi_path, soundfont_path, output_path):
    pm = pretty_midi.PrettyMIDI(midi_path)
    audio_data = pm.fluidsynth(fs=44100, synthesizer=soundfont_path)
    wavfile.write(f'{output_path}.wav', 44100, audio_data)
    return Audio(audio_data, rate=44100)
```

```python
midi_dir = Path('./babyslakh_16k/Track00001/MIDI')
test_midi = list(midi_dir.glob('*.mid'))[0]

render_midi_to_wav(str(test_midi), './FluidR3Mono_GM.sf3', 'sample_output_audio')
```

```python
model = ChordModel()
chord_annotation = model.predict(filename='sample_output_audio.wav')
df_chords = chord_annotation.to_dataframe()
df_chords.head()
```

```python
# save crema output
df_chords.to_csv('sample_output.csv', index=False)
```
